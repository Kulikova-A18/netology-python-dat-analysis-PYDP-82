Создайте новый столбец is_low_satisfaction, который равен True, если Customer Satisfaction Rating меньше 3.0, и False в противном случае.

Создайте столбец age_group по правилу:

*   Если Customer Age < 30 → значение «Молодой»;
*   Если Customer Age < 50 → значение «Средний»;

4.3. Топ самых проблемных продуктов

С помощью .value_counts() определите топ-5 продуктов (Product Purchased), на которые поступило больше всего обращений. Выведите результат.



4.1. Средние показатели

Средняя оценка по всем данным

```python
avg_overall = df['Customer Satisfaction Rating'].mean()
print(f"Средняя удовлетворенность (все данные): {avg_overall:.2f}")
```

Средняя оценка для Critical

```python
avg_critical = df[df['Ticket Priority'] == 'Critical']['Customer Satisfaction Rating'].mean()
print(f"Средняя удовлетворенность (Critical): {avg_critical:.2f}")
```

Средняя оценка для Low

```python
avg_low = df[df['Ticket Priority'] == 'Low']['Customer Satisfaction Rating'].mean()
print(f"Средняя удовлетворенность (Low): {avg_low:.2f}")
```

Средний возраст клиента

```python
avg_age = df['Customer Age'].mean()
print(f"Средний возраст клиента: {avg_age:.1f} лет")
```

4.2. Сравнение по каналам

```python
channel_avg = df.groupby('Ticket Channel')['Customer Satisfaction Rating'].mean()
print(channel_avg.sort_values(ascending=False).to_string())
```

4.3. Топ-5 самых проблемных продуктов

```python
top5_products = df['Product Purchased'].value_counts().head(5)
print("Топ-5 продуктов по количеству обращений:")
print(top5_products)
```
